```python
import yaml
import pyperclip

path = '/Users/karthik/Documents/Github/Colab/Resume.yaml'

# 1. Read and parse the YAML file into a Python object
with open(path, 'r') as f:
    resume = yaml.safe_load(f)

# # 2. Pretty-print it back out as a YAML-formatted string
# print(f"Resume: {yaml.dump(resume, default_flow_style=False, sort_keys=False)}, Job: {job}")


# Assuming 'resume' and 'job' are already defined
output_text = f"Resume: {yaml.dump(resume, default_flow_style=False, sort_keys=False)} \n Job: {job}"

# Print it to the console
# print(output_text)

# Copy it to your clipboard
pyperclip.copy(output_text)
```
