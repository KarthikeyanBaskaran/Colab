````python
#Response from groq

def llmresponsegroq(prompt):
    try:
        client = Groq(api_key=GROQ_API_KEY)
        completion = client.chat.completions.create(
            model="qwen/qwen3.8-27b",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.6,
            max_completion_tokens=1000,
            top_p=0.95,
            reasoning_effort="none",
            stream=False,
            #stream_options={"include_usage": True} # Ensures usage metadata is sent in the stream
        )
        return completion
    except Exception as e:
          logging.error(f"An error occurred while communicating with the Grok API: {e}")
          return ""
````
